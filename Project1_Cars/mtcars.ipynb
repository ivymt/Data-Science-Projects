### Exploratory Data Analysis of the mtcars dataset from Kaggle (https://www.kaggle.com/code/mmrayhan/mtcars/input)

#### The process od EDA is divided into  5 Stages:
* 1. Knowing Initial details about data
* 2. Modifying or removing unwanted data
* 3. Retrieving data
* 4. Getting statistical information 
* 5. Drawing graphs/plots

### 1. Knowing initial details about data

### 2. Modifying or Removing Unwanted Data

```python
cars= cars.drop_duplicates(keep='first')
```

```python
cars.isnull().sum()
```

```python
cars_sort= cars.sort_values(by='mpg', ascending=False)
```

```python
cars_sort
```

```python
cars['model'].unique()
```

### 3. Retrieving Data

### 4. Getting statistical information 

```python
cars1.describe()
```

```python
cars1.corr(method='pearson')
```

```python
cars1.corr(method='pearson')['mpg']
```

#### 5. Drawing Graphs/Plots

```python
sns.displot(data=cars1 , x='mpg', bins=10,kde=True, color='blue')
```

##### From the plot, we can understand thatthe highest number of cars(around 6) are giving a mileage of 15 to 20 miles per gallon. A smaller number of cars (around1) are giving mileage of 25 to 30 per gallon 

```python
sns.regplot(x='mpg', y='wt', data=cars1, color='blue')
```

##### The regression line shows negative relationship between mpg and wt. When the mpg increases the wt decreases

##### What l learned from the analysis is that the cars with higher mpg have lower weight. The cars with lower mpg have higher weight. This is a negative correlation between mpg and weight.
