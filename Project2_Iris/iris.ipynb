#### Applying SVM Model to classify the Iris Flower. Suppose we are given 3 groups of flowers called Setosa, Versicolor and Virginica.These 3 flowes belong to the same genre known as 'iris' and hence look similar in color and structure.


### Working on SVM  
* Support Vector Machine(SVM) model is implemented in a class called SVC (Support Vector Classifier)

* parameters in SVM are :
- 1. gamma value: decides whether the point close to the hyperplane should be considered to classify a data point or not. Default is 'scale'.
- 2. Regularization: helps to solve overfitting in machine learning.Shown by 'C' in SVM.
- 3. Kernel function: is a mathematical formula that is useful to rearrange the data points such that they can be divided clearly by the SVM.  The default kernel used by SCV is "rbf" radial basis function.
